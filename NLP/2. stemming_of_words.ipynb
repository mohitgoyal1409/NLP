### Stemming

Stemming is the process of cutting off the ends of words to reduce them to their base or root form (called a "stem").It is a crude, fast shortcut used in NLP to group different forms of the same word together. For example, a stemmer chops the words "running", "runs", and "ran" down to the single root stem: "run".

Stemmers do not understand human grammar or dictionary definitions; they just follow rigid, automated rules to slice characters off the end of a string. Because of this, they often cause two major issues:Over-stemming (Chops too much): It can cut different words down to the exact same incorrect root.Example: "universal", "university", and "universe" might all get aggressively chopped down to "univers".Under-stemming (Chops too little): It might fail to see that two words are related if they are spelled differently.Example: It won't recognize that "feet" is the plural form of "foot" because it can't handle irregular word changes.

```python
words = [
    'eating', 
    'eats', 
    'eaten', 
    'writing', 
    'writes', 
    'programming', 
    'programs', 
    'history', 
    'finally', 
    'finalized'

]
```

### Porterstemmer

```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
for word in words:
    print(word + '-->' + stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
stemming.stem('sitting')
```

### RegexpStemmer class

NLTK has RegexpStemmer class with the help of which we can easily implement Regular
Expression Stemmer algorithms. It basically takes a single regular expression and removes any
prefix or suffix that matches the expression. Let us see an example

```python
from nltk.stem import RegexpStemmer
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4) ## dollar in last meaning remove that word if it is occuring in last 
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

### Snowball stemmer

```python
from nltk.stem import SnowballStemmer

snowballsstemmer=SnowballStemmer('english')

for word in words:
    print(word+"-->"+snowballsstemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballsstemmer.stem("fairly"),snowballsstemmer.stem("sportingly")
```
